```python

import torch
from diffusers import StableDiffusionPipeline
from IPython.display import display

model_id = "stable-diffusion-v1-5/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    safety_checker=None
).to("cuda")

print("AI photo model loaded successfully!")
```

```python

prompt = "A beautiful red evening gown on a mannequin, elegant American fashion style, realistic fabric details, professional studio photography, plain white background"

image = pipe(
    prompt,
    num_inference_steps=20,
    guidance_scale=7.0,
    height=512,
    width=512
).images[0]

display(image)
image.save("my_first_ai_image.png")

print("Photo generated successfully!")
```

```python

import gradio as gr
import torch

def generate_photo(prompt):
    if not prompt or not prompt.strip():
        raise gr.Error("Pehle apna photo prompt likhein.")

    result = pipe(
        prompt.strip(),
        num_inference_steps=20,
        guidance_scale=7.0,
        height=512,
        width=512
    ).images[0]

    return result

with gr.Blocks(title="Zevora AI") as app:
    gr.Markdown("# Zevora AI")
    gr.Markdown("Create your own AI photos")

    prompt = gr.Textbox(
        label="Describe your image",
        placeholder="A beautiful red dress, studio photography...",
        lines=3
    )

    generate = gr.Button("Generate Photo")
    output = gr.Image(label="Your AI Photo", type="pil")

    generate.click(
        fn=generate_photo,
        inputs=prompt,
        outputs=output
    )

app.launch(share=True)
```
